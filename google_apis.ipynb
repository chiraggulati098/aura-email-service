## Gmail API use - List emails

```python
from gmail_api import init_gmail_service, get_email_messages, get_email_message_details

client_file = 'client_secret.json'
service = init_gmail_service(client_file)
```

```python
messages = get_email_messages(service, max_results = 5) # can also use 'folder_name' and 'label_ids' parameters here
```

```python
print(messages)
```

```python
for msg in messages:
    details = get_email_message_details(service, msg['id'])
    if details:
        print(details)
        print()
```
