## List emails

```python
from gmail_api import init_gmail_service, get_email_messages, get_email_message_details

client_file = 'client_secret.json'
service = init_gmail_service(client_file)
```

```python
messages = get_email_messages(service, max_results = 10) # can also use 'folder_name' and 'label_ids' parameters here
```

```python
print(messages)
```

```python
for msg in messages:
    details = get_email_message_details(service, msg['id'])
    if details:
        print(msg['id'])
        print(details)
        print()
```

## Send Email

```python
from gmail_api import send_email
```

```python
response_email_sent = send_email(
    service,
    to = "chirag.gulati@s.amity.edu",
    subject = "Test sample email",
    body = """
This is a test email.
Let's see how this goes.
    
Regards,
Chirag
    """,
    body_type='plain',
    attachment_paths=None # add attachment paths in a list
    )

print(response_email_sent)
```
